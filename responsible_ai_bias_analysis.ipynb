## Task 2.3: Propose Individual Fairness Method

1. Identifying a Sensitive/Protected Attribute
In my dataset skin tone is one of the most important sensitive attributes that could cause bias. My project classifies skin types (normal, dry, oily, sensitive and diseased skin), so I need to make sure the model does not favor one skin tone over another. If the dataset has more light skin images than dark ones, the model may work worse for some users. That would give wrong predictions for the skin tones that are underrepresented, which is unfair and not reliable.

2. Chosen Individual Fairness Method: "Fairness Through Awareness"
To deal with skin tone bias I chose the "Fairness Through Awareness" method. This approach accepts that bias exists and does something about it instead of ignoring it. Rather than pretending all images are equal, the model is made aware of skin tone differences and is trained fairly across the groups. By being aware of skin tone, the model can be adjusted to give accurate and balanced predictions for both lighter and darker skin.

3. Why "Fairness Through Awareness" works for my dataset
I think this method fits for a few reasons. First, it helps me find and fix imbalances in the dataset, because the model is trained on different skin tones. Without that step the model could just learn the bias in the data and do badly on underrepresented groups. Second, it helps avoid misclassifying skin diseases, which matters because some conditions look different on lighter and darker skin. By being aware of these differences the model can be adjusted to make fair predictions for all groups.

Another reason is that it improves generalization. If the model only sees one skin tone, it may not recognize the same condition on another skin tone. By keeping the representation more equal, the accuracy goes up for all users and the app is more reliable. Unlike "Fairness Through Unawareness", which ignores skin tone and just assumes the data is fair, this method actually deals with the bias and corrects it.
